**extra credit question 1**
<br>
1. use the quick select algorithm to partition around the median. this takes o(n) time and it would make it so the median would have less values on the left side, and greater values on the right side of the median.
2. put the smaller elements (left side of median) in the even indexes and larger elements (right side of median) on the odd elements. this would take o(n) time as well.
<br>
<br>
example:
<br>
arr = [3, 7, 1, 9, 4]
<br>
pivot = 4 (median)
<br>
smaller = [3, 1]
<br>
greater = [7, 9]
<br>
with even and odd alterating: [3, 7, 1, 9, 4]
<br>
time complexity: o(n) from partition, o(n) for placing elements = o(n)

**extra credit question 2**
<br>
in order to identify people who don't report all of their income, i would first sort both data sets using merge sort because the time complexity o(nlogn) is decent for larger data sets and is better than o(n^2). i would then linearly go through both sorted lists and compare if each report matches what the IRS actually receive. this would take o(n) time.
<br>
this would take o(nlogn), which is much more efficient than having to individually go through each report, which would be o(n^2).
