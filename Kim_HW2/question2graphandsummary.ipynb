```python
import matplotlib.pyplot as plt

# dataset
dataset_sizes = [1000, 10000, 100000, 250000, 500000, 1000000]

# elapsed time values based on dataset sizes
insertion_times = [0.016, 1.72, 198.96, 1306.05, 5666.80, 27513.09]
merge_times = [0.001, 0.02, 0.23, 0.62, 1.38, 2.97]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# insertion sort graph
ax1.plot(dataset_sizes, insertion_times, 'ro-', linewidth=2, markersize=8)
ax1.set_xlabel('data set size (no. of elements)')
ax1.set_ylabel('execution time (seconds)')
ax1.set_title('insertion sort performance - time complexity o(n^2)')
ax1.grid(True, alpha=0.3)
ax1.set_xticks(dataset_sizes)
ax1.set_xticklabels(['1K', '10K', '100K', '250K', '500K', '1M'])

# merge sort graph  
ax2.plot(dataset_sizes, merge_times, 'bo-', linewidth=2, markersize=8)
ax2.set_xlabel('data set size (no. of elements)')
ax2.set_ylabel('execution time (seconds)')
ax2.set_title('merge sort performance - time complexity o(nlogn)')
ax2.grid(True, alpha=0.3)
ax2.set_xticks(dataset_sizes)
ax2.set_xticklabels(['1K', '10K', '100K', '250K', '500K', '1M'])

# show the graphs
plt.show()
```

**both lines together**

```python
import matplotlib.pyplot as plt

# dataset
dataset_sizes = [1000, 10000, 100000, 250000, 500000, 1000000]

# elapsed time values based on dataset sizes
insertion_times = [0.016, 1.72, 198.96, 1306.05, 5666.80, 27513.09]
merge_times = [0.001, 0.02, 0.23, 0.62, 1.38, 2.97]

# creating the singular graph
plt.figure(figsize=(12, 8))
plt.plot(dataset_sizes, insertion_times, 'ro-', label='insertion sort', linewidth=2, markersize=8)
plt.plot(dataset_sizes, merge_times, 'bo-', label='merge sort', linewidth=2, markersize=8)
plt.xlabel('dataset size (no. of elements)')
plt.ylabel('execution time (seconds)')
plt.title('insertion sort vs merge sort comparison')
plt.legend()
plt.grid(True, alpha=0.3)

# Fix the axis scaling
plt.xticks(dataset_sizes, ['1K', '10K', '100K', '250K', '500K', '1M'])

plt.show()
```

this definitely shows the difference of the time complexities. the insertion sort's time complexity is o(n^2). it makes it so each element has to be compared with every other one, and it makes it tremendously show for larger data sets. merge sort on the other hand, has the time complexity o(nlogn) and was able to handle large data sets really well, only 3 seconds to complete. the merge sort is great because of it's divide and conquer strategy. it reduces the number of comparisons required since it's splitting up the work. it goes to show that sorting algorithms are fine if the data set is small, but if you increase the dataset size significantly you have to be careful which one you're going to use.
