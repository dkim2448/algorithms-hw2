# question 3: loop invariant proof
---

**3.1 general description of loop invariant technique**
<br>
a loop invariant is a property that doesn't change each time a loop iterates. it's a way to prove the correctness of the algorithm and has to be true at initialization, maintenance, and termination.

**3.2 identify loop invariant of loop in merge function**
<br>
arr[left..k-1] contains elements from arrays L and R in a SORTED order
<br>
<br>
*note:*
<br>
***arr** = the array we're sorting*
<br>
***left** = starting index of the section that we're merging*
<br>
***i** = current position of array L*
<br>
***j** = current position of array R*
<br>
***k** = the current position in arr where we're gonna place the next element (it starts at 'left')*

**3.3 initialization step**
<br>
k = left, i = 0, j = 0
<br>
arr[left..k-1] is empty because k - 1 < left. there are zero elements, so it's still sorted. thus, invariant holds initialization

**3.4 maintenance step**
<br>
when comparing L[i] and R[j], we place the smaller element into arr[k]. both L and R are sorted, so the element is the smallest among all the other elements that have yet to be compared. adding it to arr[left..k-1] makes sure that it's in sorted order and we have the smallest elements being processed each iteration. thus, the invariant holds maintenance.

**3.5 termination step**
<br>
the loop terminates when i = n1 or j = n2. this means that arr[left..k-1] contains all of the processed elements in sorted order. the rest of the elements that haven't been processed are copied to the end of arr, which still upholds that sorted property because L and R are still sorted as well. arr[left..right] is merged in a sorted order. thus, the invariant holds termination.
